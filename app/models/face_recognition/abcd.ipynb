```python
import onnx

model = onnx.load("w600k_r50.onnx")

print("Producer:", model.producer_name)
print("Version:", model.producer_version)
print("Graph:", model.graph.name)
```

```python
import onnxruntime as ort

session = ort.InferenceSession("w600k_r50.onnx")

print("Inputs:")
for i in session.get_inputs():
    print(i.name, i.shape, i.type)

print("\nOutputs:")
for o in session.get_outputs():
    print(o.name, o.shape, o.type)
```
