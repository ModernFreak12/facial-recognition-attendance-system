```python
from facenet_pytorch import InceptionResnetV1

# This downloads weights from PyTorch's CDN which ALWAYS works on Windows
model = InceptionResnetV1(pretrained='vggface2').eval()
```

```python
import torch 

model.eval()
dummy = torch.randn(1, 3, 160, 160)
torch.onnx.export(
    model,
    dummy,
    "arcface.onnx",
    input_names=["input"],
    output_names=["embedding"],
    opset_version=11
)
```
